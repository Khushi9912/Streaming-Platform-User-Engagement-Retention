```python


import pandas as pd
import csv
from io import StringIO

with open(csv_path, "r", encoding="utf-8-sig") as file:
    raw_text = file.read()

# Remove the outer quotes surrounding each complete line
fixed_text = "\n".join(
    line[1:-1] if line.startswith('"') and line.endswith('"') else line
    for line in raw_text.splitlines()
)

watch_df = pd.read_csv(StringIO(fixed_text))

print("Shape:", watch_df.shape)
print("Columns:", watch_df.columns.tolist())
print("December 31 records:",
      (watch_df["watch_date"] == "2025-12-31").sum())
```

```python

import pandas as pd
import os

# Project paths
project_path = r"C:\Users\ACER\OneDrive\PROJECTS\Khushi_Work\SQL_PROJECTS\Streaming_Platform_User_Engagement_Retention"
sql_folder = os.path.join(project_path, "04_SQL")
sql_file = os.path.join(sql_folder, "05_insert_watch_history.sql")

# Ensure numeric columns have correct types
watch_df["watch_duration_minutes"] = pd.to_numeric(
    watch_df["watch_duration_minutes"]
)
watch_df["completion_percentage"] = pd.to_numeric(
    watch_df["completion_percentage"]
)

columns = [
    "watch_id", "user_id", "content_id", "watch_date",
    "watch_duration_minutes", "completion_percentage", "device_type"
]

def sql_value(value):
    if pd.isna(value):
        return "NULL"
    if isinstance(value, str):
        return "'" + value.replace("'", "''") + "'"
    return str(value)

with open(sql_file, "w", encoding="utf-8") as f:
    f.write("USE StreamingPlatformDB;\nGO\n\n")

    for start in range(0, len(watch_df), 500):
        batch = watch_df.iloc[start:start + 500]

        f.write(
            "INSERT INTO dbo.WatchHistory ("
            + ", ".join(columns)
            + ")\nVALUES\n"
        )

        rows = []
        for _, row in batch[columns].iterrows():
            values = [sql_value(v) for v in row]
            rows.append("(" + ", ".join(values) + ")")

        f.write(",\n".join(rows) + ";\nGO\n\n")

print("SQL file generated successfully!")
print("File:", sql_file)
print("Records prepared:", len(watch_df))
```
