```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
```

```python
num_users = 2000
```

```python
user_ids = [f"U{i:05d}" for i in range(1, num_users + 1)]
user_ids
```

```python
# Generate random signup dates
```

```python
start_date = pd.Timestamp("2025-01-01")
end_date = pd.Timestamp("2025-12-31")

signup_dates = pd.to_datetime(
    np.random.randint(
        start_date.value // 10**9,
        end_date.value // 10**9,
        num_users
    ),
    unit="s"
)
```

```python
print("\nSample Signup Dates:")
print(signup_dates[:10])
```

```python
signup_dates = signup_dates.normalize()
```

```python
print("\nSignup Dates After Cleaning:")
print(signup_dates[:10])
```

```python
# Generate user countries
countries = [
    "United States",
    "India",
    "United Kingdom",
    "Canada",
    "Australia",
    "Germany",
    "France",
    "Brazil",
    "Mexico",
    "Japan"
]

country_probabilities = [
    0.30,
    0.20,
    0.10,
    0.08,
    0.07,
    0.06,
    0.05,
    0.05,
    0.05,
    0.04
]

user_countries = np.random.choice(
    countries,
    size=num_users,
    p=country_probabilities
)
```

```python
print("\nSample User Countries:")
print(user_countries[:10])
```

```python
print("\nCountry Distribution:")
print(pd.Series(user_countries).value_counts())
```

```python
# Generate subscription plans
```

```python
plans = [
    "Basic",
    "Standard",
    "Premium"
]

plan_probabilities = [
    0.40,
    0.40,
    0.20
]

subscription_plans = np.random.choice(
    plans,
    size=num_users,
    p=plan_probabilities
)
```

```python
print("\nSubscription Plan Distribution:")
print(pd.Series(subscription_plans).value_counts())
```

```python
# Generate subscription status
```

```python
statuses = [
    "Active",
    "Cancelled"
]

status_probabilities = [
    0.85,
    0.15
]

subscription_status = np.random.choice(
    statuses,
    size=num_users,
    p=status_probabilities
)
```

```python
print("\nSubscription Status Distribution:")
print(pd.Series(subscription_status).value_counts())
```

```python
# Create Users DataFrame
```

```python
users_df = pd.DataFrame({
    "user_id": user_ids,
    "signup_date": signup_dates,
    "country": user_countries,
    "subscription_plan": subscription_plans,
    "subscription_status": subscription_status
})
```

```python
print("\nUsers DataFrame:")
print(users_df.head(10))
```

```python
print("\nShape:")
print(users_df.shape)
```

```python
print("\nDuplicate User IDs:")
print(users_df["user_id"].duplicated().sum())
```

```python
print("\nMissing Values:")
print(users_df.isnull().sum())
```

```python
print("\nUnique Subscription Plans:")
print(users_df["subscription_plan"].unique())
```

```python
print("\nUnique Subscription Status:")
print(users_df["subscription_status"].unique())
```

```python
# Save Users dataset
users_df.to_csv(
    "users.csv",
    index=False
)

print("\nUsers dataset saved successfully!")
```
