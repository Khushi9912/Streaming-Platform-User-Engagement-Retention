```python
import pandas as pd
import os

project_path = r"C:\Users\ACER\OneDrive\PROJECTS\Khushi_Work\SQL_PROJECTS\Streaming_Platform_User_Engagement_Retention"

raw_file = os.path.join(
    project_path, "02_Data", "Raw", "netflix_titles.csv"
)

df = pd.read_csv(raw_file)

print("Original data shape:", df.shape)
print(df.head())
```

```python
# Make a copy of the original dataset
content_df = df.copy()

# Fill missing values
content_df["country"] = content_df["country"].fillna("Unknown")
content_df["rating"] = content_df["rating"].fillna("Unknown")

# Split duration into value and unit
content_df[["duration_value", "duration_unit"]] = (
    content_df["duration"].str.split(" ", n=1, expand=True)
)

# Convert duration to a number
content_df["duration_value"] = pd.to_numeric(
    content_df["duration_value"], errors="coerce"
).astype("Int64")

# Standardize duration units
content_df["duration_unit"] = content_df["duration_unit"].replace({
    "min": "Minutes",
    "Season": "Seasons"
})

# Clean genre formatting
content_df["listed_in"] = content_df["listed_in"].str.strip()
content_df["listed_in"] = content_df["listed_in"].str.replace(
    r"\s*,\s*", ", ", regex=True
)

# Clean country formatting
content_df["country"] = content_df["country"].str.strip()
content_df["country"] = content_df["country"].str.replace(
    r"\s*,\s*", ", ", regex=True
)

# Remove unnecessary columns
content_df = content_df.drop(
    columns=["show_id", "cast", "date_added", "description", "duration"]
)

# Create content IDs
content_df["content_id"] = [
    f"C{i:05d}" for i in range(1, len(content_df) + 1)
]

# Arrange the columns
content_df = content_df[
    [
        "content_id",
        "title",
        "type",
        "director",
        "country",
        "release_year",
        "rating",
        "duration_value",
        "duration_unit",
        "listed_in"
    ]
]

# Check the result
print("Shape:", content_df.shape)
print("Columns:", content_df.columns.tolist())
print(content_df.head())
```

```python
import os
import pandas as pd

project_path = r"C:\Users\ACER\OneDrive\PROJECTS\Khushi_Work\SQL_PROJECTS\Streaming_Platform_User_Engagement_Retention"
sql_file = os.path.join(project_path, "04_SQL", "04_insert_content.sql")

def sql_value(value):
    if pd.isna(value):
        return "NULL"
    if isinstance(value, str):
        return "'" + value.replace("'", "''") + "'"
    return str(value)

columns = [
    "content_id", "title", "type", "director", "country",
    "release_year", "rating", "duration_value",
    "duration_unit", "listed_in"
]

with open(sql_file, "w", encoding="utf-8") as f:
    f.write("USE StreamingPlatformDB;\nGO\n\n")

    for start in range(0, len(content_df), 500):
        batch = content_df.iloc[start:start + 500]

        f.write("INSERT INTO dbo.Content (" + ", ".join(columns) + ")\nVALUES\n")

        rows = []
        for _, row in batch[columns].iterrows():
            values = [sql_value(row[col]) for col in columns]
            rows.append("(" + ", ".join(values) + ")")

        f.write(",\n".join(rows) + ";\nGO\n\n")

print("SQL import file created successfully!")
print("File:", sql_file)
print("Content records:", len(content_df))
```

```python
import os
import pandas as pd

project_path = r"C:\Users\ACER\OneDrive\PROJECTS\Khushi_Work\SQL_PROJECTS\Streaming_Platform_User_Engagement_Retention"
sql_file = os.path.join(project_path, "04_SQL", "04_insert_content.sql")

# Map the DataFrame column to the SQL Server column
sql_columns = [
    "content_id", "title", "type", "director", "country",
    "release_year", "rating", "duration_value", "duration_unit", "genre"
]

# Create a SQL-safe value
def sql_value(value):
    if pd.isna(value):
        return "NULL"
    if isinstance(value, str):
        return "'" + value.replace("'", "''") + "'"
    return str(value)

# Rename listed_in to genre for SQL output
sql_df = content_df.rename(columns={"listed_in": "genre"})

with open(sql_file, "w", encoding="utf-8") as f:
    f.write("USE StreamingPlatformDB;\nGO\n\n")

    for start in range(0, len(sql_df), 500):
        batch = sql_df.iloc[start:start + 500]

        f.write(
            "INSERT INTO dbo.Content (" +
            ", ".join(sql_columns) +
            ")\nVALUES\n"
        )

        rows = []
        for _, row in batch[sql_columns].iterrows():
            values = [sql_value(row[col]) for col in sql_columns]
            rows.append("(" + ", ".join(values) + ")")

        f.write(",\n".join(rows) + ";\nGO\n\n")

print("Corrected SQL file created!")
print("Content records:", len(sql_df))
print("SQL column:", sql_columns[-1])
```
